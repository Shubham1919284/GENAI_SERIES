```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(
    model="openai/gpt-4o",
    api_key=os.getenv("OPENROUTER_GPT_API_KEY"), 
    base_url="https://openrouter.ai/api/v1",     
    max_tokens=500,  # REDUCED: Use fewer tokens to avoid credit issues
)
```

```python
text="Hello my namae is Shubham Kumar Jha, Email is sk1919284@gmail.com, age is 21 years old, I live in India, my phone number is +91-XXXXXXXXXX."
```

```python
# FIXED: Use fewer tokens to avoid credit issues
res=llm.invoke(f"give me only name, email, age, phone, location:{text}")
# print('Response:', res.content if hasattr(res, 'content') else res)
print(res.content)
```

### Structured Output

```python
from pydantic import BaseModel, Field
from typing import Optional

class StructuredResponse(BaseModel):
    name: str = Field(..., description="User Name")
    email: str = Field(..., description="User Email")
    age: int = Field(..., description="User Age")
    phone: str = Field(..., description="User Phone Number")
    location: str = Field(..., description="User Location")
StructuredResponse=llm.with_structured_output(StructuredResponse)
```

```python
res=StructuredResponse.invoke(f"give me only name, email, age, phone, location:{text}")
```

```python
class movies(BaseModel):
    title:str=Field(...,description="Movies Title")
    genre:str=Field(...,description="Movies Genre")
    year:int=Field(...,description="Release Year ")
    rating:float=Field(...,description="IMDB Rating")

class Recommendation(movies):
    movies:list[movies]

movies=llm.with_structured_output(Recommendation)
```

```python
res=movies.invoke("Give me details of the recent coming movies")
```
