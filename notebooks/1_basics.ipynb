### Connect with Gemini Model


```python
from dotenv import load_dotenv
import os

# Load environment variables from .env file
load_dotenv('.env')

# Verify the key is loaded
google_api_key = os.getenv('GOOGLE_API_KEY') or os.getenv('Google_API_KEY')
print("Google API key loaded:", bool(google_api_key))
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

# Gemini uses the Google API key from the .env file
gemini_llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    google_api_key=google_api_key
)
```

```python
from langchain_groq import ChatGroq
groq_llm = ChatGroq(
    model="llama-3.1-8b-instant",
    api_key=os.getenv("GROQ_API_KEY"),
    max_tokens=1000
)
```

```python
res = groq_llm.invoke("Explain the theory of relativity in simple terms.")
print(res.content)
```

```python
# CORRECT WAY: Using LangChain message classes
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Explain the theory of relativity in simple terms.")
]

res = groq_llm.invoke(messages)
print("With System Message:")
print(res.content)
print("\n" + "="*50)

```

```python
# Alternative: Using dictionary format (also works)
messages_dict = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Explain quantum computing in simple terms."}
]

res2 = groq_llm.invoke(messages_dict)
print("With Dictionary Format:")
print(res2.content)
```
