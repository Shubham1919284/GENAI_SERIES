```python
from dotenv import load_dotenv
load_dotenv('.env')
import os
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    api_key=os.getenv("GROQ_API"),  
)
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int):
    """
        it will return the sum of two numbers

        Args:
            a : Number One
            b : Number Two
    """
    return a+b

@tool
def multiply_numbers(a:int, b:int):
    """
        it will return the multiply of two numbers

        Args:
            a : Number One
            b : Number Two
    """
    return a*b
```

```python
add_numbers.invoke({"a":12, "b":3})
```

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are a math teacher. Always use tools for calculation. After getting the result from a tool, always explain the answer in a complete sentence."
)

response = agent.invoke({"messages": [{"role": "user", "content": "what is 5*9?"}]})
print(response["messages"][-1].content)
```
