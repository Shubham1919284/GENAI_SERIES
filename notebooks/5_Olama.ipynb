### Implementing the Olama Model :

```python
from langchain_ollama import ChatOllama

# Initialize Ollama model (requires Ollama to be running locally)
# Make sure to run: ollama serve
# And pull a model: ollama pull llama2

llm = ChatOllama(
    model="llama2",  # or any model you have installed locally
    temperature=0.7
)

# Test the model
response = llm.invoke("Hello! Can you tell me about yourself?")
print("Ollama Response:", response.content)
```

```python
history=[]
while True:
    query=input("User: ")
    if query.lower() in ["exit","bye","quit"]:  
        print("GoodBye 👋")
        break
    
    history.append({"role":"user","content":query})
    print("User: ",query)

    res=llm.invoke(history)
    history.append({"role":"ai", "content":res.content})
    print("AI: ",res.content,"\n")
```
